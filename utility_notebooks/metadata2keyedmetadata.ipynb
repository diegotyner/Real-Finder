```python
import json
import os

INPUT_FILE = "songs_metadata.json"
OUTPUT_FILE = "keyed_songs_metadata.json"

def create_edition_index(master_data):

    edition_index = {
        "5": [],
        "6": []
    }

    for song in master_data:
        if not song.get('title'):
            continue

        for location in song.get('found', []):
            edition = str(location.get('edition_found'))
            if edition in edition_index:
                entry = {
                    "title": song['title'],
                    "scanned_title": location.get('scanned_title', 'Unknown'),
                    "composer": song.get('composer', 'Unknown'),
                    "page_number": location.get('page_number'),
                    "key": song.get('key', None)
                }
                if entry['page_number'] is not None:
                    edition_index[edition].append(entry)
    for edition in edition_index:
        edition_index[edition].sort(key=lambda x: x['page_number'])

    return edition_index

def main():
    with open(INPUT_FILE, 'r', encoding='utf-8') as f:
        master_data = json.load(f)

    print(f"Loaded {len(master_data)} master song entries.")

    transformed_data = create_edition_index(master_data)

    with open(OUTPUT_FILE, 'w', encoding='utf-8') as f:
        json.dump(transformed_data, f, indent=2, ensure_ascii=False)

    print(f"Index created successfully:")
    print(f"  Edition 5 songs: {len(transformed_data['5'])}")
    print(f"  Edition 6 songs: {len(transformed_data['6'])}")
    print(f"Saved to: {OUTPUT_FILE}")


if __name__ == "__main__":
    main()
```
